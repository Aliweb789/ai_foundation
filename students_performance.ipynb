```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Student_Performance.csv")

print("Dataset shape (rows, columns):", df.shape)
print("Column name:", df.columns.tolist())

print("First few rows of the dataset:")
print(df.head())

print("Summary statistics:")
print(df.describe())

print("Data types of columns:")
print(df.info())
```

```python
print("Average score by columns:")
df[["math_score", "english_score", "science_score"]].mean()
```

```python
print("Maximum score by subject:")
df[["math_score", "english_score", "science_score"]].max()
```

```python
print("Minimum score by columns:")
df[["math_score", "english_score", "science_score"]].min()
```

```python
# Correlation
print("Correlation:")
df[["math_score", "english_score", "science_score"]].corr()
```

```python
df["total score"] = df[["math_score", "english_score", "science_score"]].sum(axis=1)
df["average score:"] = df["total score"] / 3
df["passed"] = np.where(df["average score:"] >= 50, "Pass", "Fail")

print("New coulmns preview (total, average, passed):")
print(df[["total score", "average score:", "passed"]].head())
```

```python
test_prep_group = df.groupby("study_method")[["math_score", "english_score", "science_score"]].mean()
print("Students score by test preparation course:")
print(test_prep_group)

# Gender-based average
gender_group = df.groupby("gender")[["math_score", "english_score", "science_score"]].mean()
print("Average score by gender:")
print(gender_group)

# Who passed vs failed
pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate:")
print(pass_rate)
```

```python
# Convert score to NumPy array
scores = df[["math_score", "english_score", "science_score"]].values

# Mean, Std Deviation
print("NumPy Mean:", np.mean(scores, axis=0))
print("NumPy Std:", np.std(scores, axis=0))

# Z-score (normalize)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("Z-scores (normalized numbers)")
print(z_score[:5])
# Average 0
# Negative -
# Positive +
```

```python
# How many students scored above 90 average?
top_scores = df[df["average score:"] >= 90]
print("Top scored students above 90:", len(top_scores))

# How many failed?
failed = df[df["passed"] == "Fail"]
print("Failed students:", len(failed))
```
