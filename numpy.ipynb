```python
import numpy as np
arr = np.array([1, 3, 5, 7])
print("numpy_array:", arr)
arr.shape
```

```python
arr2 = np.array([[1, 3, 5, 7], [3, 5, 6, 9]])
print("numpy_array2:", arr2)
arr2.shape
```

```python
# dtype()
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape()
a = np.array([1, 2, 3, 4, 5, 6,])
b = a.reshape(2, 3)
print("shaped_array", b)
```

```python
# astype() typeni ozgartirib beradi
a = np.array([1, 2, 3])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeros(), np.ones() => dummy data
z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros:", z)
print("ones:", o)
```

```python
# ndim() --> demention ni aniqlaydi
a = np.array([[1,2], [3, 4]])
print("ndim:", a.ndim)
```

```python
# arange() --> oraligidagi sonlarni chiqaradi(test data uchun)
a = np.arange(1, 7)
print("arange:", a)
```

```python
# sum(), mean() --> sonlar ortachasini aniqlaydi, max(), min()
a = np.array([1, 2, 3, 4])

print("sum:", a.sum())
print("mean:", a.mean())
print("max:", a.max())
print("min:", a.min())
```

```python
# flatten() -->1D ga otkazib beradi(vektorga otkazib beradi)
a = np.array([[1, 2], [3, 4]])
print('flatten:', a.flatten())
```

```python
# T (Transpose) ->
a = np.array([[1, 2], [3, 4]])
print("transpose:", a.T)
```

```python
# dot() @--> kopaytirish vaqtida ishlatiladi
a = np.array([[1,2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("dot/multiplication:", a.dot(b))
```

```python
# random.rand() 0.0 and 1.0 => 0.0 included but 1.0 excluded
a = np.random.rand(2, 3)
print("random rand:", a)
```
